Montée en CFL – loi géométrique

Génère `Tab_Params`, `Tab_ValuesParams` et `Tab_Iter` au format attendu par le solveur.

```python
ite_begin, ite_end = 5, 2000
cfl_begin, cfl_end = 1.0, 200.0
every = 10  # mise à jour tous les combien d'itérations

ratio = cfl_end / cfl_begin

step_iters = list(range(ite_begin, ite_end, every)) + [ite_end]
step_cfls = [
    cfl_begin * ratio ** ((it - ite_begin) / (ite_end - ite_begin))
    for it in step_iters
]

PARAM_NAMES = ["CFL number", "CFL number (coarse grids)"]

Tab_Params = [name for _ in step_cfls for name in PARAM_NAMES]
Tab_ValuesParams = [str(round(cfl, 6)) for cfl in step_cfls for _ in PARAM_NAMES]
Tab_Iter = [it for it in step_iters for _ in PARAM_NAMES]

print(f'Tab_Params = {Tab_Params}')
print(f'Tab_ValuesParams = {Tab_ValuesParams}')
print(f'Tab_Iter = {Tab_Iter}')
```

Plot

```python
import plotting
import matplotlib.pyplot as plt

plotting.use_style("paper")

fig, ax = plt.subplots()
plotting.plot_line(ax, step_iters, step_cfls)

ax.set_xlabel("ITE")
ax.set_ylabel("CFL")
ax.set_title("CFL vs ITE avec une loi géométrique")
ax.legend()
plt.show()
```
